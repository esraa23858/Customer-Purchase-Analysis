```python
##Task 4. Exploratory Analysis
```

```python

df["Category"].value_counts()
```

```python

# Analysis 1: Frequency of product categories
category_frequency = df["Category"].value_counts()

print("Purchase Records by Category:")
print(category_frequency)

category_percentage = (category_frequency / len(df) * 100).round(2)
print("\nCategory Percentages:")
print(category_percentage)
```

```python
# Analysis 2: Total purchase amount by category
category_sales = (
    df.groupby("Category")["Purchase Amount (USD)"]
    .sum()
    .sort_values(ascending=False)
)

print("Total Purchase Amount by Category:")
print(category_sales)
```

```python

# Analysis 3: Average purchase amount by season and category
season_category = (
    df.groupby(["Season", "Category"])["Purchase Amount (USD)"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

print("Average Purchase Amount by Season and Category:")
print(season_category)

print("\nHighest Average:")
print(season_category.head(1))

print("\nLowest Average:")
print(season_category.tail(1))
```

```python
# Analysis 4: Average review rating by category
category_rating = (
    df.groupby("Category")["Review Rating"]
    .mean()
    .round(2)
    .sort_values(ascending=False)
)

print("Average Review Rating by Category:")
print(category_rating)

print("\nHighest Rated Category:")
print(category_rating.head(1))

print("\nLowest Rated Category:")
print(category_rating.tail(1))
```

```python
##Task 5. Findings
```

```python
# Finding 1: Product Category Frequency

category_counts = df["Category"].value_counts()
category_percentages = (df["Category"].value_counts(normalize=True) * 100).round(2)

category_summary = pd.DataFrame({
    "Count": category_counts,
    "Percentage (%)": category_percentages
})

print(category_summary)
```

```python
# Finding 2: Gender Distribution

gender_counts = df["Gender"].value_counts()
gender_percentages = (df["Gender"].value_counts(normalize=True) * 100).round(2)

gender_summary = pd.DataFrame({
    "Count": gender_counts,
    "Percentage (%)": gender_percentages
})

print(gender_summary)
```

```python
# Finding 3: Purchase Amount Statistics

purchase_summary = df["Purchase Amount (USD)"].agg(
    ["count", "mean", "min", "max"]
).round(2)

print(purchase_summary)
```
